```python
import os, glob
import pandas as pd
import gseapy as gp
```

```python
pth = os.path.join('..', '..', 'data')
pth_feat = os.path.join(pth, 'features', 'biomart')
pth_out = os.path.join(pth, 'results', 'gseapy')
```

```python
feat_fn = sorted(glob.glob(os.path.join(pth_feat, '*.csv')))
get_src = lambda fn: os.path.split(fn)[1].replace('.csv', '')
feat_dict = {get_src(fn) : fn for fn in feat_fn if 'union' not in fn}

thresh = .05
for src, fn in feat_dict.items():
    print(src)
    g = pd.read_csv(fn).mmusculus.drop_duplicates()

    # GO term enrichment analysis
    enr = gp.enrichr(
            gene_list = g,
            gene_sets = 'GO_Biological_Process_2025',
            organism = 'mouse',
            outdir = None)
    df = enr.results.copy()
    df = df.loc[df['Adjusted P-value'] < thresh].copy()
    df.to_csv(os.path.join(pth_out, f'{src}.csv'))
```
