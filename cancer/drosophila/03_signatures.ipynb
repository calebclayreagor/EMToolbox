```python
import os, glob
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
```

```python
pth = os.path.join('..', '..')
pth_data = os.path.join(pth, 'data')
pth_in = os.path.join(pth_data, 'cancer', 'drosophila')
pth_feat = os.path.join(pth_data, 'features', 'biomart')
pth_out = os.path.join(pth, 'figures', 'drosophila')
```

```python
adata = sc.read_h5ad(os.path.join(pth_in, 'follicle.h5ad'))
adata
```

```python
fn = os.path.join(pth_data, 'summary.csv')
df_summary = pd.read_csv(fn, index_col = 0)
labels_dict = df_summary.Signature.dropna().to_dict()
labels_dict
```

```python
feat_fn = sorted(glob.glob(os.path.join(pth_feat, '*.csv')))
get_src = lambda fn: os.path.split(fn)[1].replace('.csv', '')
feat_dict = {get_src(fn) : pd.read_csv(fn) for fn in feat_fn}

# module signatures
g_ref = feat_dict['union'].dmelanogaster.drop_duplicates()
g_ref = g_ref.loc[g_ref.isin(adata.var_names)]
for src, df_src in feat_dict.items():
    if src != 'union':
        print(src)
        g = df_src.dmelanogaster.drop_duplicates()
        g = g.loc[g.isin(adata.var_names)]
        sc.tl.score_genes(adata,
                          gene_list = g,
                          score_name = src,
                          ctrl_as_ref = False,
                          gene_pool = g_ref,
                          random_state = 1234)
```

```python
ix = ['Mitotic',
      'Postmitotic',
      'Main Body',
      'Posterior',
      'Terminal',
      'Stretched',
      'Border',
      'Stalk']
```

```python
# signatures dotplot
fs1, fs2, src = 9, 8, list(labels_dict.keys())
dp = sc.pl.dotplot(adata, src, 'celltype',
                   categories_order = ix,
                   dot_max = .8,
                   vmin = 0,
                   cmap = 'plasma',
                   figsize = (1.85, 2),
                   return_fig = True,
                   show = False)

dp.style(largest_dot = 120,
         x_padding = .7,
         y_padding = .6)
dp.legend(width = .9)

ax = dp.get_axes()
ax_main = ax['mainplot_ax']
ax_size = ax['size_legend_ax']
cax = ax['color_legend_ax']

ax_size.set_title('% Cells', size = fs2, pad = -10)
ax_size.tick_params('x', size = 0, labelsize = fs2, pad = 2.5)
ax_size.set_xticklabels(['20', '40', '60', '   ≥80'])
cax.set_title('Signature', size = fs2, pad = 6)
cax.set_xticks([0, .05], ['0', '0.05'])
cax.tick_params('x', size = 0, labelsize = fs2, pad = 6)

pos_main = ax_main.get_position()
pos_size = ax_size.get_position()
pos_color = cax.get_position()

ax_size.set_position([pos_main.x0 + .65 * pos_main.width,
                      pos_main.y0 - .925 * pos_main.height,
                      pos_main.width,
                      pos_size.height])

cax.set_position([pos_main.x0 + .1 * pos_main.width,
                  pos_main.y0 - .56 * pos_main.height,
                  .8 * pos_main.width,
                  pos_color.height])

ax_main.set_xticklabels([labels_dict[lbl.get_text()] for
                         lbl in ax_main.get_xticklabels()],
                        fontsize = fs1)
ax_main.set_yticklabels([])
ax_main.tick_params('y', size = 2.5)

fn = os.path.join(pth_out, 'signatures.pdf')
plt.savefig(fn, bbox_inches = 'tight')
```
