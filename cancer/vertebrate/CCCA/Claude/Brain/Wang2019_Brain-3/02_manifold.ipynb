# Wang et al. 2019 — Glioblastoma (GBM)
## 02 · Manifold & trajectory

Fit a scFates principal graph to the 10x malignant manifold. **The geometry (branching vs
non-branching) is decided by biology** — whether Gast and pSom form two competing arms — not by an
ElPiGraph node count (04_test attempts 3/5 failure: topology read from a metric, not the embedding).
We score the three developmental modules, graph-smooth them, read their spatial relationship, fit a
**curve** (the correct topology for a collapsed single axis), verify it is **fold-free**, root it at
the **Gast/pSom-high** developmental origin (Crest is *late*, never the root), and confirm continuity
on an embedding the curve was **not** fit on (the Kürten fold failure mode).

```python
import os, sys, glob, io, contextlib, warnings; warnings.filterwarnings('ignore')
os.environ['R_HOME'] = sys.exec_prefix + '/lib/R/'
%matplotlib inline
import numpy as np, pandas as pd
import scanpy as sc, scanpy.external as sce, scFates as scf
import anndata as ad
import matplotlib.pyplot as plt
from sklearn.neighbors import NearestNeighbors
from scipy.sparse.csgraph import connected_components, shortest_path
from scipy.stats import spearmanr
import scipy.sparse as sp
sc.settings.verbosity = 1
SEED = 1; np.random.seed(SEED)
```

```python
# cwd = <root>/cancer/vertebrate/CCCA/Claude/<Category>/<Dataset>  (two levels below Puram)
name = os.path.basename(os.getcwd())                    # 'Wang2019_Brain-3'
cat  = os.path.basename(os.path.dirname(os.getcwd()))   # 'Brain'
pth  = os.path.join(*([os.pardir]*6))                   # -> MesenCoder root
pth_data = os.path.join(pth, 'data')
sys.path.append(os.path.join(pth, 'cancer', 'vertebrate', 'CCCA'))
from scripts.load import load_CCCA_adata
# raw input folder = dataset minus the attempt suffix and with the Data_ prefix
raw = 'Data_' + name.split('-')[0]                      # 'Data_Wang2019_Brain'
pth_in  = os.path.join(pth_data, 'unzip', 'CCCA', raw)
pth_out = os.path.join(pth_data, 'cancer', 'vertebrate', 'CCCA', 'Claude', cat, name)
pth_fig = os.path.join(pth, 'figures', 'cancer', 'vertebrate', 'CCCA', 'Claude', cat, name)
pth_feat= os.path.join(pth_data, 'features', 'biomart')
pth_cb  = os.path.join(pth_fig, 'colorbars')
os.makedirs(pth_out, exist_ok=True); os.makedirs(pth_cb, exist_ok=True)
```

### Load processed manifold; score + graph-smooth developmental modules
`score_genes` for Gast/pSom/Crest against the union control pool, then kNN-connectivity smoothing so geography is read from a de-noised field rather than per-cell noise (the raw scores remain for the 03 deliverable panels).

```python
adata = sc.read_h5ad(os.path.join(pth_out, 'processed.h5ad'))
print('whole malignant manifold:', adata.shape[0], 'cells')
xy = adata.obsm['X_umap']
from numpy.linalg import svd
cc = xy - xy.mean(0); u, sv, vt = svd(cc, full_matrices=False)
print(f'UMAP aspect ratio (major/minor std): {sv[0]/sv[1]:.2f}')

feat = {os.path.basename(fn).replace('.csv',''): pd.read_csv(fn).hsapiens.drop_duplicates()
        for fn in sorted(glob.glob(os.path.join(pth_feat,'*.csv')))}
g_ref = feat['union'].loc[feat['union'].isin(adata.var_names)]
label = {'GSE162534':'Gast','GSE229103':'pSom','rRNAModifications':'Crest'}
for src, lab in label.items():
    g = feat[src].loc[feat[src].isin(adata.var_names)]
    sc.tl.score_genes(adata, gene_list=g, score_name=lab, ctrl_as_ref=False,
                      gene_pool=g_ref, random_state=1234)
    print(f'{lab}: {g.size}/{feat[src].size} genes present')
A = (adata.obsp['connectivities']>0).astype(float); A = A + sp.eye(A.shape[0])
A = sp.diags(1/np.array(A.sum(1)).ravel()) @ A
S = adata.obs[['Gast','pSom','Crest']].values.copy()
for _ in range(3): S = A @ S
for j, lab in enumerate(['Gast','pSom','Crest']): adata.obs[lab+'_sm'] = S[:,j]
```

### Geometry from biology — is there a Gast↔pSom cross-repression axis?
A **branching** case needs Gast and pSom to *anti*-correlate spatially (two competing arms). If they
**co-vary** and both oppose Crest, the programs are present but collapsed onto one axis — the
**non-branching** case (§1) — and a curve is the correct topology. We also confirm proliferation is
orthogonal to the developmental axis.

```python
rr = {}
print('smoothed spatial correlations on the whole manifold:')
for a, b in [('Gast_sm','pSom_sm'),('Crest_sm','Gast_sm'),('Crest_sm','pSom_sm')]:
    rr[(a,b)] = np.corrcoef(adata.obs[a], adata.obs[b])[0,1]; print(f'  {a} vs {b}: r = {rr[(a,b)]:+.3f}')
r_gp = rr[('Gast_sm','pSom_sm')]
# geometry from biology: co-varying Gast/pSom (r>0) => collapsed single axis (non-branching);
# spatially anti-correlated Gast/pSom (r<0) => two competing arms (branching).
geometry = 'NON-BRANCHING' if r_gp > 0 else 'BRANCHING'
print(f'\nGast/pSom spatial r = {r_gp:+.3f} -> {geometry} '
      + ('(co-vary; both oppose Crest -> single collapsed axis).' if r_gp>0 else '(anti-correlate -> two arms).'))

cyc_orth = {}
print('\nproliferation orthogonality (corr of computed cyc_score with smoothed modules):')
for m in ['Gast_sm','pSom_sm','Crest_sm']:
    cyc_orth[m] = np.corrcoef(adata.obs.cyc_score, adata.obs[m])[0,1]
    print(f'  cyc_score vs {m}: r = {cyc_orth[m]:+.3f}')
orthogonal = max(abs(v) for v in cyc_orth.values()) < 0.2
print(f'\nper-compartment smoothed developmental means:')
print(adata.obs.groupby('compartment', observed=True)[['Gast_sm','pSom_sm','Crest_sm']].mean().round(4).to_string())
hi = adata.obs['Crest_sm'] >= adata.obs['Crest_sm'].quantile(0.9)
frac = adata.obs.loc[hi.values,'compartment'].value_counts(normalize=True).round(2).to_dict()
print('\nCrest-high decile across compartments:', frac)
print(f'-> proliferation orthogonal to the developmental axis: {orthogonal} (max |r| < 0.2); '
      + ('keep whole manifold (no scoping).' if orthogonal else 'scope out cycling lobe.'))
assert geometry == 'NON-BRANCHING', 'geometry check: expected co-varying Gast/pSom on this dataset'
sub = adata   # whole manifold is the trajectory object
```

### Fit the curve and verify it is fold-free
A curve is only valid if it does not fold back through the cloud. `fold_diag` reports tips, forks, and
the minimum self-approach of graph-distant nodes relative to the median edge length (>1 = fold-free).
We assert zero forks and no fold before trusting the pseudotime.

```python
def fold_diag(ad):
    B = np.asarray(ad.uns['graph']['B']); F = ad.uns['graph']['F'].T
    G = shortest_path(sp.csr_matrix((B>0).astype(int)), unweighted=True)
    ii, jj = np.where(np.triu(B>0)); med_edge = np.median(np.sqrt(((F[ii]-F[jj])**2).sum(1)))
    dmin = np.inf
    for a in range(F.shape[0]):
        for b in range(a+1, F.shape[0]):
            if G[a,b] >= 3:
                dmin = min(dmin, np.sqrt(((F[a]-F[b])**2).sum()))
    tips = int(((B>0).sum(0)==1).sum()); forks = int(((B>0).sum(0)>=3).sum())
    return dmin/med_edge, tips, forks

with contextlib.redirect_stdout(io.StringIO()):
    scf.tl.curve(sub, Nodes=20, use_rep='X_umap', epg_lambda=0.3, epg_mu=100.0, seed=SEED, plot=False)
sa, tips, forks = fold_diag(sub)
print(f'curve: {tips} tips, {forks} forks | fold self-approach ratio = {sa:.2f} (>1 = fold-free)')
assert forks == 0 and sa > 1, 'topology check failed'
```

### Root from developmental order, compute pseudotime
Score each tip by its smoothed module means and root at the **Gast/pSom-high, Crest-low** tip — the developmental origin. Crest is a *late* readout, so it is never the root.

```python
B = np.asarray(sub.uns['graph']['B']); tps = np.where((B>0).sum(0)==1)[0]
R = np.asarray(sub.obsm['X_R']); hard = R.argmax(1)
D = shortest_path(sp.csr_matrix((B>0).astype(int)), unweighted=True)
info = {}
for tp in tps:
    m = np.isin(hard, np.where(D[tp]<=3)[0])
    info[tp] = (sub.obs['Gast_sm'][m].mean(), sub.obs['pSom_sm'][m].mean(), sub.obs['Crest_sm'][m].mean())
    print(f'tip {tp}: Gast={info[tp][0]:+.4f} pSom={info[tp][1]:+.4f} Crest={info[tp][2]:+.4f}')
root = int(max(tps, key=lambda tp: info[tp][0]+info[tp][1]-info[tp][2]))
print('derived root = tip', root, '(Gast/pSom-high, Crest-low developmental origin)')
with contextlib.redirect_stdout(io.StringIO()):
    scf.tl.root(sub, root); scf.tl.pseudotime(sub, n_map=1, seed=SEED)
t = sub.obs['t'].values
print('\ngradient along pseudotime (Spearman rho on smoothed scores):')
for sname in ['Gast_sm','pSom_sm','Crest_sm','cyc_score']:
    rho, p = spearmanr(t, sub.obs[sname]); print(f'  {sname}: rho={rho:+.3f} p={p:.1e}')
```

### Continuity off an independent embedding
The curve was fit on the UMAP, so continuity cannot be judged with a UMAP-neighbour metric (blind to a fold in that same UMAP — the Kürten failure mode). We check pseudotime against a **diffusion-map** component the curve never touched.

```python
sc.tl.diffmap(sub, n_comps=15)
dc1 = sub.obsm['X_diffmap'][:,1]
rho_dc, _ = spearmanr(t, dc1)
print(f'continuity: corr(pseudotime, diffusion DC1) = {rho_dc:+.2f} on an independent embedding')
tn = (t-t.min())/(t.max()-t.min())
hi = sub.obs['Crest_sm'] >= sub.obs['Crest_sm'].quantile(0.9)
print(f'Crest-high decile mean pseudotime = {tn[hi.values].mean():.2f} (terminal if ~1)')
print('monotonic on an embedding the curve was not fit on -> continuous, fold-free confirmed.')
```

### Orient to the house linear layout (pseudotime left → right)

```python
xy = sub.obsm['X_umap'].copy(); t = sub.obs['t'].values
xyr = np.column_stack([xy[:,0], xy[:,1]])
r = float(np.corrcoef(xyr[:,0], t)[0,1])
if r < 0:
    xyr[:,0] = -xyr[:,0]; r = -r
assert r > 0, f'pseudotime not left->right (corr={r:+.2f})'
sub.obsm['X_umap_lin'] = xyr
print(f'orientation: corr(x_lin, pseudotime) = {r:+.2f} (low pt -> left)')
```

### Render the fitted curve on the embedding (catch folds visually)

```python
F = np.asarray(sub.uns['graph']['F']); Bc = np.asarray(sub.uns['graph']['B'])
ii, jj = np.where(np.triu(Bc>0))
fig, ax = plt.subplots(figsize=(4.6, 3.6))
sca = ax.scatter(sub.obsm['X_umap'][:,0], sub.obsm['X_umap'][:,1], c=t, cmap='cividis', s=6, alpha=.6)
for a, b in zip(ii, jj):
    ax.plot([F[0,a],F[0,b]], [F[1,a],F[1,b]], '-', color='crimson', lw=2, zorder=5)
ax.scatter(F[0], F[1], c='crimson', s=18, zorder=6)
ax.scatter(F[0,root], F[1,root], c='lime', s=140, marker='*', zorder=7, label='root')
ax.axis('off'); ax.legend(); ax.set_title('Fitted curve on 10x manifold (pseudotime)', size=9)
plt.colorbar(sca, ax=ax, label='pseudotime'); plt.show()
```

### Rename milestones + deliverable `manifold.pdf` / `manifold.h5ad`

```python
# name the two tips by developmental role
scf.tl.rename_milestones(sub, new={str(root): 'Origin_GastpSom', str(int([tp for tp in tps if tp!=root][0])): 'Terminus_Crest'})
# sanitize non-serializable ElPiGraph uns entry
sub.uns['epg'] = {'NodePositions': np.asarray(sub.uns['epg']['NodePositions'])}
sub.write(os.path.join(pth_out, 'manifold.h5ad'))
xl = sub.obsm['X_umap_lin']
xlo, xhi = np.percentile(xl[:,0], [0.3,99.7]); ylo, yhi = np.percentile(xl[:,1], [0.3,99.7])
px = (xhi-xlo)*.05; py = (yhi-ylo)*.08
plot = sub.copy(); plot.obsm['X_umap'] = plot.obsm['X_umap_lin']
fig, ax = plt.subplots(figsize=(3.5,1.5))
sc.pl.umap(plot, color='t', cmap='cividis', size=30, alpha=.8, colorbar_loc='right',
           frameon=False, show=False, ax=ax, title='Pseudotime')
ax.set_xlim(xlo-px, xhi+px); ax.set_ylim(ylo-py, yhi+py)
ax.text(.5, -.12, 'Wang et al., 2019', ha='center', transform=ax.transAxes, size=6)
fig.savefig(os.path.join(pth_fig,'manifold.pdf'), bbox_inches='tight'); plt.show()
print('saved manifold.h5ad + manifold.pdf')
```
