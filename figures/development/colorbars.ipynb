```python
import os
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from matplotlib.colorbar import ColorbarBase
```

```python
for cmap, label, pth in (
    ('viridis', 'PseudoTime', 'trajectories'),
    ('RdBu_r', 'Scaled\nExpression', 'features'),
    ('plasma', 'Scaled\nSignature', 'signatures')):

    fig = plt.figure(figsize = (.66, 1))
    cax = fig.add_axes([.35, .05, .3, .9])
    norm = Normalize(vmin = 0, vmax = 1)
    cb = ColorbarBase(cax, cmap = cmap, norm = norm)
    cb.set_ticks([0, 1], labels = ['Min', 'Max'])
    cb.ax.tick_params(size = 0)
    cb.set_label(label)
    fn = os.path.join(pth, 'colorbar.pdf')
    plt.savefig(fn, bbox_inches = 'tight')
```
