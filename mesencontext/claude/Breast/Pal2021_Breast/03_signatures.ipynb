# Pal et al. 2021 — Breast Cancer (10x)
## 03 · Embryonic module signatures — **Gate 2**

Render the raw per-cell Gast / pSom / Crest signatures on the manifold and test the module geography.
**Gate 2** asks whether the embryonic pattern appears. Here Gast/pSom do not resolve into two competing
arms; instead a single embryonic axis runs pSom-high (origin) → Crest-high (terminus), with an **inverse
pSom↔Crest gradient** — the non-branching case (§1), which passes Gate 2 (GO) with Gate 3 = NA.

```python
import os, sys, glob
os.environ['R_HOME'] = sys.exec_prefix + '/lib/R/'
import numpy as np, pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
from scipy.stats import spearmanr
sc.settings.verbosity = 1
SEED = 1; np.random.seed(SEED)
```

```python
name = os.path.basename(os.getcwd())
cat  = os.path.basename(os.path.dirname(os.getcwd()))
pth  = os.path.join(*([os.pardir]*4))
pth_data = os.path.join(pth, 'data')
pth_out = os.path.join(pth_data, 'cancer', 'vertebrate', 'CCCA', 'claude', cat, name)
pth_fig = os.path.join(pth, 'figures', 'mesencontext', 'claude', cat, name)
pth_cb  = os.path.join(pth_fig, 'colorbars')
os.makedirs(pth_out, exist_ok=True); os.makedirs(pth_cb, exist_ok=True)
```

```python
adata = sc.read_h5ad(os.path.join(pth_out, 'manifold.h5ad'))
modules = {'Gast':'Gastrulation','pSom':'Somitogenesis','Crest':'Neural Crest'}
print('cells:', adata.shape[0], '| Body cells with pseudotime:', int(adata.obs['t'].notna().sum()))
pth_feat = os.path.join(pth_data, 'features', 'biomart')
feat = {os.path.basename(f).replace('.csv',''): pd.read_csv(f).hsapiens.drop_duplicates()
        for f in sorted(glob.glob(os.path.join(pth_feat,'*.csv')))}
for src,lab in [('GSE162534','Gast'),('GSE229103','pSom'),('rRNAModifications','Crest')]:
    g = feat[src].loc[feat[src].isin(adata.var_names)]; print(f'{lab}: {g.size}/{feat[src].size} genes present')
```

### Gate 2 — module geography
Smoothed spatial correlations and the pseudotime gradient over the Body. The bifurcation signature would
be a Gast/pSom hub diverging into two arms with Gast↔pSom mutual exclusivity *and* a Crest tip. Instead:
pSom and Crest are strongly **anti-correlated** (single inverse gradient), Gast is weak and tracks mildly
with pseudotime, and pSom&Crest joint-high enrichment is < 1 (they avoid each other — an inverse gradient,
not a co-expressing hub). Single axis, no bifurcation → **Gate 2 GO (non-branching); Gate 3 NA**.

```python
print('smoothed spatial correlations (whole manifold):')
for a,b in [('Gast_sm','pSom_sm'),('Crest_sm','Gast_sm'),('Crest_sm','pSom_sm')]:
    print(f'  {a} vs {b}: r = {np.corrcoef(adata.obs[a],adata.obs[b])[0,1]:+.2f}')
body = adata[adata.obs['t'].notna()].copy()
t = body.obs['t'].values; tn=(t-t.min())/(t.max()-t.min())
print('gradient along pseudotime over the Body (Spearman rho on smoothed):')
for s in ['Gast_sm','pSom_sm','Crest_sm']:
    rho,_=spearmanr(t, body.obs[s]); print(f'  {s}: rho = {rho:+.3f}')
hi = body.obs['Crest_sm'] >= body.obs['Crest_sm'].quantile(0.9)
print(f'Neural-crest-high decile mean pseudotime = {tn[hi.values].mean():.2f} (terminal if ~1)')
pz=(body.obs['pSom']-body.obs['pSom'].mean())/body.obs['pSom'].std()
cz=(body.obs['Crest']-body.obs['Crest'].mean())/body.obs['Crest'].std()
joint=((pz>0.5)&(cz>0.5)).mean(); exp=(pz>0.5).mean()*(cz>0.5).mean()
print(f'joint pSom&Crest-high enrichment = {joint/exp:.2f} (<1 = inverse gradient, not a hub)')
print('Gate 2: single (pSom/Gast)->Crest gradient + inverse terminal Crest -> GO (non-branching)')
print('Gate 3: no bifurcation -> NA')
```

### Figures — raw per-cell signatures (deliverables)
The deliverable panels show **raw** per-cell `score_genes` values (smoothing aids the analysis above but
the panels show the honest raw read), on the shared house linear layout with external colorbars.

```python
plot = adata.copy(); plot.obsm['X_umap'] = plot.obsm['X_umap_lin']
xl = plot.obsm['X_umap']
xlo,xhi=np.percentile(xl[:,0],[0.3,99.7]); ylo,yhi=np.percentile(xl[:,1],[0.3,99.7])
px=(xhi-xlo)*.05; py=(yhi-ylo)*.08
titles={'Gast':'Gastrulation Signature','pSom':'Somitogenesis Signature','Crest':'Neural Crest Signature'}
for m in ['Gast','pSom','Crest']:
    fig, ax = plt.subplots(figsize=(3.5,1.5))
    sc.pl.umap(plot, color=m, cmap='plasma', alpha=.7, size=8, colorbar_loc=None,
               frameon=False, show=False, ax=ax, title=titles[m])
    ax.set_xlim(xlo-px,xhi+px); ax.set_ylim(ylo-py,yhi+py)
    ax.text(.5,-.12,'Pal et al., 2021', ha='center', transform=ax.transAxes, size=6)
    fig.savefig(os.path.join(pth_fig, f'{m}_signature.pdf'), bbox_inches='tight')
    lo,hi_=float(adata.obs[m].min()),float(adata.obs[m].max())
    ticks=np.round(np.linspace(lo,hi_,5),2)
    fcb=plt.figure(figsize=(.66,1)); cax=fcb.add_axes([.35,.05,.3,.9])
    cb=plt.colorbar(ax.collections[0], cax=cax); cb.ax.tick_params(which='both', size=0)
    cb.set_label(f'{m}\nSignature'); cb.set_ticks(ticks, labels=ticks); cb.solids.set_alpha(1)
    fcb.savefig(os.path.join(pth_cb, f'{m}_colorbar.pdf'), bbox_inches='tight')
    print('saved', f'{m}_signature.pdf', '+ colorbar')
```
