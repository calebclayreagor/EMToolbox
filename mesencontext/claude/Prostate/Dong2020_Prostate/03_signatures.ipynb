# 03 · Signatures — embryonic-development module scores & Gate 2

The three embryonic modules (**Gast** gastrulation, **pSom** posterior somitogenesis, **Crest**
neural crest) were scored in 02 against the shared **union** control pool and graph-smoothed.
Here we render them as deliverables and evaluate **Gate 2**.

**Gate 2 (non-branching case).** Puram's branching manifold splits Gast and pSom into *distinct
arms* with Crest terminal and cross-repressing. A non-branching manifold instead shows a **single
developmental gradient** with an inverse terminal end. In Dong 2020, Gast and pSom **co-vary**
(smoothed r ≈ +0.90; joint Gast&pSom-high enrichment ≈ 2.4, an overlap — *not* two competing
arms), and this shared origin programme is opposed to **neural-crest** at the terminus (pSom↔Crest
smoothed r ≈ −0.53). Along pseudotime Gast and pSom decline (rho ≈ −0.58 / −0.71) while Crest
rises to a terminal end (rho ≈ +0.79; Crest-high decile at pseudotime 0.98). The single-axis,
overlap-origin / inverse-terminal pattern **is present**, so **Gate 2 = GO**; with no bifurcation
the cross-repression test is **Gate 3 = NA**.

**The gate is geography, not means.** The raw module means here (Gast +0.035 / pSom +0.100 /
Crest −0.041) are pSom-dominant — the same profile as many lung / PDAC runs. What decides GO vs
NO-GO is the *geography*: a coherent Gast/pSom→Crest axis with Crest genuinely terminal passes,
whereas mutually-exclusive module territories with no terminal readout do not. Deliverable:
`signatures.h5ad`, `{Gast,pSom,Crest}_signature.pdf` + colorbars.

```python
import os, sys, glob
os.environ['R_HOME'] = sys.exec_prefix + '/lib/R/'
import numpy as np, pandas as pd
import scanpy as sc, scanpy.external as sce, scFates as scf
import matplotlib.pyplot as plt
from scipy.stats import spearmanr
sc.settings.verbosity = 1
SEED = 1; np.random.seed(SEED)
```

```python
name = os.path.basename(os.getcwd())
cat  = os.path.basename(os.path.dirname(os.getcwd()))
pth  = os.path.join(*([os.pardir]*4))
pth_data = os.path.join(pth, 'data')
pth_out = os.path.join(pth_data, 'cancer', 'vertebrate', 'CCCA', 'claude', cat, name)
pth_fig = os.path.join(pth, 'figures', 'mesencontext', 'claude', cat, name)
pth_cb  = os.path.join(pth_fig, 'colorbars')
os.makedirs(pth_out, exist_ok=True); os.makedirs(pth_cb, exist_ok=True)
```

### Load the rooted manifold (signatures scored + smoothed in 02)

```python
adata = sc.read_h5ad(os.path.join(pth_out, 'manifold.h5ad'))
modules = {'Gast':'Gastrulation', 'pSom':'Somitogenesis', 'Crest':'Neural Crest'}
print('cells:', adata.shape[0], '| scores present:', [m for m in modules if m in adata.obs])
pth_feat = os.path.join(pth_data, 'features', 'biomart')
feat = {os.path.basename(f).replace('.csv',''): pd.read_csv(f).hsapiens.drop_duplicates()
        for f in sorted(glob.glob(os.path.join(pth_feat,'*.csv')))}
for src,lab in [('GSE162534','Gast'),('GSE229103','pSom'),('rRNAModifications','Crest')]:
    g = feat[src].loc[feat[src].isin(adata.var_names)]
    print(f'{lab}: {g.size}/{feat[src].size} genes present')
```

### Gate 2 — module geography on the manifold
Quantitative test of the non-branching signature for this dataset: (i) Gast and pSom co-vary
(overlap, not competing arms) and jointly oppose Crest; (ii) along pseudotime Gast/pSom decline
and Crest rises (Crest terminal); (iii) the joint pSom&Crest-high enrichment is a depletion (an
inverse gradient), while the joint Gast&pSom-high enrichment is >1 (a co-expression overlap).

```python
print('smoothed spatial correlations (geography):')
for a,b in [('Gast_sm','pSom_sm'),('Crest_sm','Gast_sm'),('Crest_sm','pSom_sm')]:
    r = np.corrcoef(adata.obs[a], adata.obs[b])[0,1]; print(f'  {a} vs {b}: r = {r:+.2f}')
t = adata.obs['t'].values; tn = (t-t.min())/(t.max()-t.min())
print('gradient along pseudotime (Spearman rho on smoothed):')
for s in ['Gast_sm','pSom_sm','Crest_sm']:
    rho,_ = spearmanr(t, adata.obs[s]); print(f'  {s}: rho = {rho:+.3f}')
hi = adata.obs['Crest_sm'] >= adata.obs['Crest_sm'].quantile(0.9)
print(f'Neural-crest-high decile mean pseudotime = {tn[hi.values].mean():.2f} (terminal if ~1)')
# inverse-gradient vs co-expression-hub tests
z = lambda k: (adata.obs[k]-adata.obs[k].mean())/adata.obs[k].std()
gz, pz, cz = z('Gast'), z('pSom'), z('Crest')
enr = lambda a,b: ((a>0.5)&(b>0.5)).mean() / ((a>0.5).mean()*(b>0.5).mean())
print(f'joint pSom&Crest-high enrichment = {enr(pz,cz):.2f} (<1 = inverse gradient, not a hub)')
print(f'joint Gast&pSom-high enrichment  = {enr(gz,pz):.2f} (>1 = co-expression overlap, not competing arms)')
print('Gate 2: single Gast/pSom->Crest gradient (Gast/pSom overlap, inverse terminal Crest) -> GO (non-branching)')
print('Gate 3: no bifurcation -> NA')
```

### Per-module signature figures + colorbars (deliverable)
House style: `figsize=(3.5,1.5)`, pseudotime left→right (`X_umap_lin`), `size=30`, plasma,
**raw** (not smoothed) score per cell so 10x dropout is visible. Colorbar ticks data-driven.

```python
plot = adata.copy(); plot.obsm['X_umap'] = plot.obsm['X_umap_lin']
xl = plot.obsm['X_umap']
xlo,xhi = np.percentile(xl[:,0],[0.3,99.7]); ylo,yhi = np.percentile(xl[:,1],[0.3,99.7])
px=(xhi-xlo)*.05; py=(yhi-ylo)*.08
titles = {'Gast':'Gastrulation Signature','pSom':'Somitogenesis Signature','Crest':'Neural Crest Signature'}
for m in ['Gast','pSom','Crest']:
    fig, ax = plt.subplots(figsize=(3.5,1.5))
    sc.pl.umap(plot, color=m, cmap='plasma', alpha=.7, size=30, colorbar_loc=None,
               frameon=False, show=False, ax=ax, title=titles[m])
    ax.set_xlim(xlo-px,xhi+px); ax.set_ylim(ylo-py,yhi+py)
    ax.text(.5,-.12,'Dong et al., 2020', ha='center', transform=ax.transAxes, size=6)
    fig.savefig(os.path.join(pth_fig, f'{m}_signature.pdf'), bbox_inches='tight')
    lo, hi_ = float(adata.obs[m].min()), float(adata.obs[m].max())
    ticks = np.round(np.linspace(lo, hi_, 5), 2)
    fcb = plt.figure(figsize=(.66,1)); cax = fcb.add_axes([.35,.05,.3,.9])
    cb = plt.colorbar(ax.collections[0], cax=cax); cb.ax.tick_params(which='both', size=0)
    cb.set_label(f'{m}\nSignature'); cb.set_ticks(ticks, labels=ticks); cb.solids.set_alpha(1)
    fcb.savefig(os.path.join(pth_cb, f'{m}_colorbar.pdf'), bbox_inches='tight')
    print('saved', f'{m}_signature.pdf', '+ colorbar')
```

### Save signatures data

```python
adata.write(os.path.join(pth_out, 'signatures.h5ad'))
print('saved signatures.h5ad', adata.shape)
```
