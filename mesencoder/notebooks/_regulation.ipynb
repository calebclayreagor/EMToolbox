```python
import os
import numpy as np
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
```

```python
pth = os.path.join('..', '..')
pth_data = os.path.join(pth, 'data')
pth_in = os.path.join(pth_data, 'development', 'modeling')
pth_ref = os.path.join(pth_data, 'ref', 'AnimalTFDBv4')
# pth_fig = os.path.join(pth, 'figures', 'development', 'regulation')
# pth_out = os.path.join(pth_data, 'results', 'regulation')
```

```python
adata = sc.read_h5ad(os.path.join(pth_in, 'bifurcation.h5ad'))
adata.X = np.asarray(adata.X.toarray() - adata.X.mean(axis = 1))
```

```python
adata1 = sc.read_h5ad(os.path.join(pth_in, 'competition.h5ad'))
adata1
```

```python
# transcription factors
fn = os.path.join(pth_ref, 'Mus_musculus_TF')
tf = pd.read_csv(fn, sep = '\t').Symbol
tf = tf.loc[tf.isin(adata1.var_names)]
X = adata1[:, tf].X.toarray()
msk = (X.mean(axis = 0) > 1e-5)
tf = tf.loc[msk].values
```

```python
# X: TF expression
X = adata1[:, tf].X.toarray()
X = StandardScaler().fit_transform(X)

# y: 
y = adata1.obs.competition.values.reshape(-1, 1)
y = StandardScaler().fit_transform(y).flatten()

# linear regression (y ~ X)
reg = LinearRegression().fit(X, y)

# TF coefficients (beta)
beta = pd.Series(reg.coef_, index = tf)
beta = beta.sort_values(ascending = False); beta.name = 'coef'
# beta.to_csv(os.path.join(pth_out, 'development.csv'));
beta
```

```python
vlim = {'Mesp1' : [3.25, 'p90'], 'Etv2' : []}

fig, ax = plt.subplots(1, 1, figsize = (3.5, 1.5))
sc.pl.embedding(adata, 'X_mesencoder', dimensions = [1, 0],
                size = 80, alpha = .5,
                frameon = False, show = False, ax = ax)
for g, (vmin, vmax) in vlim.items():
    sc.pl.embedding(adata[adata[:, g].X.toarray() > vmin],
                    'X_mesencoder', dimensions = [1, 0],
                    size = 80, color = g,
                    cmap = 'Reds', vmin = vmin,
                    vmax = vmax, colorbar_loc = None,
                    frameon = False, show = False, ax = ax)
ax.set_title(''); ax.invert_yaxis()

# fn = os.path.join(pth_fig, f'{geo}.pdf')
# plt.savefig(fn, bbox_inches = 'tight')
```
