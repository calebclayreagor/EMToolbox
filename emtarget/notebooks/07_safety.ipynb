```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
pth = os.path.join('..', '..')
pth_in = os.path.join(pth, 'data', 'ref', 'CELLxGENE')
pth_out = os.path.join(pth, 'figures', 'emtarget')
```

```python
# load CZ CELLxGENE gene expression (pseudobulk)
fn = os.path.join(pth_in, 'pbulk.csv')
df = pd.read_csv(fn, index_col = [0, 1])
```

```python
g = ['ANXA5', 'ARL6IP1', 'LAPTM4A',
     'PPIB', 'SPINT2', 'SPINT1', 'GPC3']

fig, ax = plt.subplots(1, 1, figsize = (2, 2.5))
sns.violinplot(df.loc[:, g], cut = 0,
               facecolor = 'lightblue',
               inner = 'quart')
ax.collections[-1].set_facecolor('salmon')
ax.tick_params('x', rotation = 90)
ax.set_ylabel('Normal Expression')
fig.suptitle('Target Safety', size = 13, y = 1.085)
fig.text(.5, .95, 'CZ CELLxGENE', ha = 'center')
for t in ax.get_xticklabels():
    t.set_fontstyle('italic')
fn = os.path.join(pth_out, 'safety.pdf')
plt.savefig(fn, bbox_inches = 'tight')
```
